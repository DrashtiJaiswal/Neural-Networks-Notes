```python
import numpy as np
print("Imported successfully")
```

```python
numbers = np.array([1, 2, 3,  4, 5])
print(numbers)
```

```python
print("Shape", numbers.shape)
print("Size", numbers.size)
print("Dimensions", numbers.ndim)
print("DataTypes", numbers.dtype)
```

```python
zeros = np.zeros((3,3))

ones = np.ones((2,4))

identity = np.eye(3)

print(zeros)
print(ones)
print(identity)
```

```python
a = np.arange(1, 11)
b = np.arange(0, 21, 2)

print(a)
print(b)
```

```python
c = np.linspace(0, 10, 5)
print(c)
```

```python
arr = np.array([10, 20, 30, 40, 50])

print(arr[1:4])
print(arr[:3])
print(arr[3:])
```

```python
# Vectorization
arr = np.array([1, 2, 3, 4])
result_np = arr * 10
print(result_np)
```

```python
# Broadcasting
matrix = np.array([10, 20, 30])
result = matrix * 2 #the scalar 2 is stretched to match[2,2,2]
print(result)
```

```python
A = np.array([
    [1,2],
    [3,4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print(np.matmul(A, B))
```
