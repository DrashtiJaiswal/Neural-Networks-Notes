```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
sales_data = {
    "Months": ["January", "February", "March", "April", "May", "June", "July", "August", "September", "October", "November", "December"],
    "Sales": [120, 135, 150, 165, 180, 210, 240, 300, 200, 310, 290, 330],
    "Profits": [25, 32, 42, 55, 71, 89, 56, 32, 96, 55, 32, 43],
    "Customers": [240, 31, 120, 159, 256, 220 , 310, 124, 290, 320, 370, 270]
    } #12 values of each

df = pd.DataFrame(sales_data)
df
```

```python
plt.figure(figsize=(10, 5))
plt.plot(df["Months"], df["Sales"], marker='o',label="Sales")
plt.plot(df["Months"], df["Profits"], marker='s',label="Profits")
plt.title("Montly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(10, 5))
plt.bar(df["Months"], df["Sales"])
plt.title("Sales Monthly")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```

```python
plt.figure(figsize=(10, 5))
plt.barh(df["Months"], df["Sales"])
plt.title("Sales Monthly")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(df["Customers"], df["Sales"], s=100)
plt.title("Customers vs Sales")
plt.xlabel("Customers")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8, 8))
plt.pie(df["Profits"],
        labels=df["Months"],
        autopct="%1.1f%%")
plt.title("Customers vs Sales")
plt.show()
```

```python
correlation = df[["Sales", "Profits", "Customers"]].corr()
plt.figure(figsize=(5, 4))
plt.imshow(correlation, cmap="coolwarm")
plt.title("Correlation Matrix/HeatMap")
plt.xticks(range(len(correlation)), correlation.columns)
plt.yticks(range(len(correlation)), correlation.columns)
plt.colorbar()
plt.show()
```
