```python
def mcp_neuron(inputs, threshold):
  total = sum(inputs)

  if total >= threshold:
    return 1

  return 0
```

```python
print(mcp_neuron([1,1],2))
print(mcp_neuron([1,0],2))
print(mcp_neuron([0,0],2))
```

```python
and_dataset = [
    [0,0],
    [0,1],
    [1,0],
    [1,1]
]

for sample in and_dataset:
  print(sample, " : ", mcp_neuron(sample,2))
```

```python
for sample in and_dataset:
  print(sample, " : ", mcp_neuron(sample,1))
```

```python
import matplotlib.pyplot as plt
class0 = [(0,0),(0,1),(1,0)]
class1 = [(1,1)]
for x,y in class0:
  plt.scatter(x,y,s=200,label="Class 0", marker="o")
for x,y in class1:
  plt.scatter(x,y,s=200,label="Class 1", marker="^")
plt.axline((0.5,1.5), (1.5,0.5), linewidth = 2)
plt.xlim(-0.5,1.5)
plt.ylim(-0.5,1.5)
plt.grid()
plt.title("AND Gate is Linearly Separable")
plt.show()
```

```python
import matplotlib.pyplot as plt
class0 = [(0,0)] #change
class1 = [(1,1) ,(0,1),(1,0)] #change
for x,y in class0:
  plt.scatter(x,y,s=200,label="Class 0", marker="o")
for x,y in class1:
  plt.scatter(x,y,s=200,label="Class 1", marker="^")
plt.axline((-0.5,0.5), (0.5,-0.5), linewidth = 2) #change
plt.xlim(-0.5,1.5)
plt.ylim(-0.5,1.5)
plt.grid()
plt.title("OR Gate is Linearly Separable")
plt.show()
```
