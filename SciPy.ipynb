Scientific Python library built on top of NumPy

```python
import pandas as pd
import numpy as np
from scipy import stats
```

```python
sales_data = {
    "Months": ["January", "February", "March", "April", "May", "June", "July", "August", "September", "October", "November", "December"],
    "Sales": [120, 135, 150, 165, 180, 210, 240, 300, 200, 310, 290, 330],
    "Profits": [25, 32, 42, 55, 71, 89, 56, 32, 96, 55, 32, 43],
    "Customers": [240, 31, 120, 159, 256, 220 , 310, 124, 290, 320, 370, 270]
    } #12 values of each

df = pd.DataFrame(sales_data)
df
```

```python
print("Mean Sales", stats.tmean(df["Sales"]))
print("Median Sales", np.median(df["Sales"]))
print("Mode Sales", stats.mode(df["Sales"]))
```

```python
print("Mean:", stats.tmean([5,9,8,12]))
print("Standard Deviation:", np.std([5,9,8,12]))
```
